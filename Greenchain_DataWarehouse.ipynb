# **ETL (Extract, Transform, Load)**

## Extract

```python
df = df.copy()

print("1. Total Baris :", df.shape[0])
print("2. Total Kolom :", df.shape[1])
print("="*25)

print("\n3. NULL VALUES")
print(df.isnull().sum())
print("="*40)

print("\n4. DUPLIKAT")
print("Jumlah duplikat:", df.duplicated().sum())
print("="*40)
```

## Transform

### Menghapus Kolom Null

```python
kolom_kosong = df.columns[df.isnull().all()].tolist()
print("Kolom dihapus:", kolom_kosong)

df = df.drop(columns=kolom_kosong)
print("Shape setelah drop:", df.shape)
```

### Rename Kolom

```python
df = df.rename(columns={
    'ID'                        : 'id',
    'Product_Type'              : 'product_type',
    'Raw_Material_Usage_kg'     : 'raw_material_usage_kg',
    'Energy_Consumption_kWh'    : 'energy_consumption_kwh',
    'Waste_Generated_kg'        : 'waste_generated_kg',
    'Transport_Distance_km'     : 'transport_distance_km',
    'CO2_Emissions_kg'          : 'co2_emissions_kg',
    'Manufacturing_Energy_kWh'  : 'manufacturing_energy_kwh',
    'Renewable_Energy_%'        : 'renewable_energy_pct',
    'Cost_$'                    : 'cost_usd',
    'Delivery_Time_days'        : 'delivery_time_days',
    'Sustainability_Score'      : 'sustainability_score',
    'Recyclability Score (0-1)' : 'recyclability_score',
    'Carbon Footprint (kg CO2)' : 'carbon_footprint_kg_co2'
})

print("Kolom setelah rename:")
print(df.columns.tolist())
```

### Handle Missing Values

```python
print("Missing values per kolom:")
print(df.isnull().sum())
```

```python
num_cols = df.select_dtypes(include='number').columns.tolist()
num_cols.remove('id')

for col in num_cols:
    missing = df[col].isnull().sum()
    if missing > 0:
        median_val = df[col].median()
        df[col].fillna(median_val, inplace=True)
        print(f"'{col}': {missing} nilai diisi median ({median_val:.2f})")

print("\nTotal missing setelah handling:", df.isnull().sum().sum())
```

### Handle Duplikat

```python
dup_count = df.duplicated(subset=df.columns.difference(['id'])).sum()
print(f"Duplikat ditemukan: {dup_count}")

df = df.drop_duplicates(subset=df.columns.difference(['id']))
print(f"Shape setelah deduplikasi: {df.shape}")
```

### Validasi Tipe Data

```python
df['product_type']        = df['product_type'].astype('category')
df['recyclability_score'] = df['recyclability_score'].astype(int)
df['id']                  = df['id'].astype(int)

print(df.dtypes)
```

### Cek Standar Deviasi

```python
# ════ CEK STANDAR DEVIASI ════
print("=" * 55)
print("ANALISIS STANDAR DEVIASI")
print("=" * 55)

num_cols = df.select_dtypes(include='number').columns.tolist()
num_cols.remove('id')

stats_std = []

for col in num_cols:
    mean  = df[col].mean()
    std   = df[col].std()
    cv    = (std / mean * 100) if mean != 0 else 0

    # Klasifikasi variabilitas berdasarkan CV
    if cv < 15:
        variabilitas = " Rendah"
    elif cv < 30:
        variabilitas = " Sedang"
    else:
        variabilitas = " Tinggi"

    stats_std.append({
        'Kolom'         : col,
        'Mean'          : round(mean, 2),
        'Std Dev'       : round(std, 2),
        'Min'           : round(df[col].min(), 2),
        'Max'           : round(df[col].max(), 2),
        'CV (%)'        : round(cv, 2),
        'Variabilitas'  : variabilitas
    })

df_std = pd.DataFrame(stats_std)
print(df_std.to_string(index=False))

print(f"\nCatatan:")
print(f"   CV < 15%  → data homogen, variasi kecil")
print(f"   CV 15–30% → variasi sedang, masih wajar")
print(f"   CV > 30%  → variasi tinggi, perlu perhatian ekstra saat outlier handling")

# Visualisasi distribusi kolom dengan std tinggi
high_cv = [row['Kolom'] for row in stats_std if row['CV (%)'] > 30]
print(f"\nKolom dengan CV tinggi (> 30%): {high_cv if high_cv else 'Tidak ada'}")

if high_cv:
    import matplotlib.pyplot as plt
    fig, axes = plt.subplots(1, len(high_cv), figsize=(5 * len(high_cv), 4))
    if len(high_cv) == 1:
        axes = [axes]
    for ax, col in zip(axes, high_cv):
        df[col].hist(bins=30, ax=ax, color='steelblue', edgecolor='white', linewidth=0.5)
        ax.set_title(f'{col}\n(CV = {round(df[col].std()/df[col].mean()*100, 1)}%)')
        ax.set_xlabel('Nilai')
        ax.set_ylabel('Frekuensi')
    plt.suptitle('Distribusi Kolom dengan Variabilitas Tinggi', fontsize=13, fontweight='bold')
    plt.tight_layout()
    plt.show()
```

### Cek Inkonsistensi Data

```python
# CEK INKONSISTENSI DATA
print("=" * 55)
print("   🔍  ANALISIS INKONSISTENSI DATA")
print("=" * 55)

inkonsistensi = []

# 1. Cek nilai negatif pada kolom yang seharusnya positif
cols_positif = [
    'raw_material_usage_kg', 'energy_consumption_kwh',
    'waste_generated_kg',    'transport_distance_km',
    'co2_emissions_kg',      'manufacturing_energy_kwh',
    'cost_usd',              'delivery_time_days',
    'sustainability_score',  'carbon_footprint_kg_co2'
]
for col in cols_positif:
    if col in df.columns:
        n = df[df[col] < 0].shape[0]
        inkonsistensi.append({
            'Kolom'   : col,
            'Jenis'   : 'Nilai Negatif',
            'Jumlah'  : n,
            'Status'  : 'OK' if n == 0 else f'! {n} baris'
        })

# 2. Cek nilai di luar rentang logis
batas = {
    'sustainability_score'  : (0, 100),
    'renewable_energy_pct'  : (0, 100),
    'recyclability_score'   : (0, 1),
    'delivery_time_days'    : (1, 30),
}
for col, (lo, hi) in batas.items():
    if col in df.columns:
        n = df[(df[col] < lo) | (df[col] > hi)].shape[0]
        inkonsistensi.append({
            'Kolom'   : col,
            'Jenis'   : f'Di luar rentang ({lo}–{hi})',
            'Jumlah'  : n,
            'Status'  : 'OK' if n == 0 else f'! {n} baris'
        })

# 3. Cek inkonsistensi kategori (spasi tersembunyi / huruf campuran)
cat_cols = df.select_dtypes(include=['object', 'category']).columns.tolist()
for col in cat_cols:
    # Cek jika ada nilai unik yang mirip karena spasi atau kapital
    vals_raw   = df[col].astype(str).unique()
    vals_clean = df[col].astype(str).str.strip().str.title().unique()
    selisih = len(vals_raw) - len(vals_clean)
    inkonsistensi.append({
        'Kolom'   : col,
        'Jenis'   : 'Inkonsistensi Kategori (spasi/kapital)',
        'Jumlah'  : selisih if selisih > 0 else 0,
        'Status'  : 'OK' if selisih <= 0 else f'!!  {selisih} variasi'
    })

# 4. Cek sustainability_category konsisten dengan sustainability_score
if 'sustainability_category' in df.columns and 'sustainability_score' in df.columns:
    def expected_cat(score):
        if score <= 33: return 'Low'
        elif score <= 66: return 'Medium'
        else: return 'High'

    df['_expected_cat'] = df['sustainability_score'].apply(expected_cat)
    mismatch = df[df['sustainability_category'].astype(str) != df['_expected_cat']].shape[0]
    df.drop(columns=['_expected_cat'], inplace=True)
    inkonsistensi.append({
        'Kolom'   : 'sustainability_category vs score',
        'Jenis'   : 'Inkonsistensi Label vs Nilai',
        'Jumlah'  : mismatch,
        'Status'  : 'OK' if mismatch == 0 else f'!!  {mismatch} baris tidak cocok'
    })

# 5. Cek apakah co2_emissions_kg dan carbon_footprint_kg_co2 tidak saling kontradiksi
if 'co2_emissions_kg' in df.columns and 'carbon_footprint_kg_co2' in df.columns:
    rasio = df['co2_emissions_kg'] / (df['carbon_footprint_kg_co2'] + 1)
    ekstrem = df[(rasio > 10) | (rasio < 0.1)].shape[0]
    inkonsistensi.append({
        'Kolom'   : 'co2 vs carbon_footprint',
        'Jenis'   : 'Rasio Ekstrem (> 10x atau < 0.1x)',
        'Jumlah'  : ekstrem,
        'Status'  : 'OK' if ekstrem == 0 else f'!  {ekstrem} baris mencurigakan'
    })

# Tampilkan hasil
df_inkon = pd.DataFrame(inkonsistensi)
print(df_inkon.to_string(index=False))

# Ringkasan
total_masalah = df_inkon[~df_inkon['Status'].str.startswith('OK')]['Jumlah'].sum()
print(f"\n{'='*55}")
if total_masalah == 0:
    print("Tidak ditemukan inkonsistensi data yang signifikan.")
else:
    print(f"Total potensi inkonsistensi: {total_masalah} baris/item")
    print(f"   → Lanjutkan ke Handle Outlier untuk penanganan lebih lanjut.")
print("=" * 55)
```

### Handle Outlier (Winsorize - IQR)

```python
cols_outlier = [
    'raw_material_usage_kg', 'energy_consumption_kwh',
    'waste_generated_kg',    'transport_distance_km',
    'co2_emissions_kg',      'manufacturing_energy_kwh',
    'cost_usd',              'sustainability_score',
    'carbon_footprint_kg_co2'
]

print("Outlier sebelum winsorize:")
for col in cols_outlier:
    Q1    = df[col].quantile(0.25)
    Q3    = df[col].quantile(0.75)
    IQR   = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    count = df[(df[col] < lower) | (df[col] > upper)].shape[0]
    print(f"  {col}: {count} outlier")
```

```python
for col in cols_outlier:
    Q1    = df[col].quantile(0.25)
    Q3    = df[col].quantile(0.75)
    IQR   = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df[col] = df[col].clip(lower=lower, upper=upper)

print("Outlier selesai di-cap (winsorized).")
```

### Validasi Logis Bisnis

```python
checks = {
    "Sustainability score di luar 0-100" : df[(df['sustainability_score'] < 0)   | (df['sustainability_score'] > 100)],
    "Recyclability bukan binary (0/1)"   : df[~df['recyclability_score'].isin([0, 1])],
    "Renewable energy pct di luar 0-100" : df[(df['renewable_energy_pct'] < 0)   | (df['renewable_energy_pct'] > 100)],
}

for pesan, hasil in checks.items():
    status = "OK" if len(hasil) == 0 else f"! {len(hasil)} baris bermasalah"
    print(f"{status}  →  {pesan}")
```

### Feature Engineering

# **DATA WAREHOUSE**

## Buat Schema (Dim & Fact Table)

```python
# ── dim_product ──
dim_product = (
    df[['product_type','sustainability_category']]
    .drop_duplicates(subset='product_type')
    .sort_values('product_type')
    .reset_index(drop=True)
)
dim_product.insert(0, 'product_key', range(1, len(dim_product)+1))
print('dim_product:')
display(dim_product)

# ── dim_delivery ──
dim_delivery = (
    df[['delivery_time_days']]
    .drop_duplicates()
    .sort_values('delivery_time_days')
    .reset_index(drop=True)
)
dim_delivery.insert(0, 'delivery_key', range(1, len(dim_delivery)+1))
print('\ndim_delivery:')
display(dim_delivery)

# ── fact_supply_chain ──
df_fact = df.merge(dim_product[['product_key','product_type']], on='product_type', how='left')
df_fact = df_fact.merge(dim_delivery[['delivery_key','delivery_time_days']], on='delivery_time_days', how='left')

fact_cols = [
    'id','product_key','delivery_key',
    'raw_material_usage_kg','energy_consumption_kwh','waste_generated_kg',
    'transport_distance_km','co2_emissions_kg','manufacturing_energy_kwh',
    'renewable_energy_pct','cost_usd','delivery_time_days',
    'sustainability_score','recyclability_score','carbon_footprint_kg_co2',
    'energy_efficiency_ratio','total_carbon_index','cost_per_km'
]
fact_table = df_fact[fact_cols].rename(columns={'id':'fact_id'}).reset_index(drop=True)

print(f'\nfact_supply_chain: {fact_table.shape[0]} baris × {fact_table.shape[1]} kolom')
print('Semua tabel siap!')
```

## Load Data ke Warehouse

```python
def val(v):
    if v is None or (isinstance(v, float) and np.isnan(v)):
        return 'NULL'
    if isinstance(v, str):
        return "'" + v.replace("'", "\\'") + "'"
    if isinstance(v, (np.integer, int)):
        return str(int(v))
    if isinstance(v, (np.floating, float)):
        return str(round(float(v), 6))
    return "'" + str(v) + "'"

def df_to_inserts(df, table_name, chunk=500):
    lines = []
    cols  = ', '.join([f'`{c}`' for c in df.columns])
    rows  = df.values.tolist()
    for i in range(0, len(rows), chunk):
        batch    = rows[i:i+chunk]
        val_strs = [f"({', '.join([val(x) for x in row])})" for row in batch]
        lines.append(f"INSERT INTO `{table_name}` ({cols}) VALUES\n" + ',\n'.join(val_strs) + ';')
    return '\n'.join(lines)

sql = []
sql.append('-- GreenChain Database')
sql.append('CREATE DATABASE IF NOT EXISTS `greenchain_db`')
sql.append('  CHARACTER SET utf8mb4 COLLATE utf8mb4_unicode_ci;')
sql.append('USE `greenchain_db`;')
sql.append('')

sql.append('DROP TABLE IF EXISTS `fact_supply_chain`;')
sql.append('DROP TABLE IF EXISTS `dim_product`;')
sql.append('DROP TABLE IF EXISTS `dim_delivery`;')
sql.append('')

sql.append("""CREATE TABLE `dim_product` (
  `product_key`             INT          NOT NULL AUTO_INCREMENT,
  `product_type`            VARCHAR(100) NOT NULL,
  `sustainability_category` VARCHAR(20)  NOT NULL,
  PRIMARY KEY (`product_key`),
  UNIQUE KEY `uq_product_type` (`product_type`)
) ENGINE=InnoDB DEFAULT CHARSET=utf8mb4;
""")
sql.append(df_to_inserts(dim_product, 'dim_product'))
sql.append('')

sql.append("""CREATE TABLE `dim_delivery` (
  `delivery_key`       INT NOT NULL AUTO_INCREMENT,
  `delivery_time_days` INT NOT NULL,
  PRIMARY KEY (`delivery_key`),
  UNIQUE KEY `uq_delivery_days` (`delivery_time_days`)
) ENGINE=InnoDB DEFAULT CHARSET=utf8mb4;
""")
sql.append(df_to_inserts(dim_delivery, 'dim_delivery'))
sql.append('')

sql.append("""CREATE TABLE `fact_supply_chain` (
  `fact_id`                  INT        NOT NULL,
  `product_key`              INT        NOT NULL,
  `delivery_key`             INT        NOT NULL,
  `raw_material_usage_kg`    DOUBLE     NOT NULL,
  `energy_consumption_kwh`   DOUBLE     NOT NULL,
  `waste_generated_kg`       DOUBLE     NOT NULL,
  `transport_distance_km`    DOUBLE     NOT NULL,
  `co2_emissions_kg`         DOUBLE     NOT NULL,
  `manufacturing_energy_kwh` DOUBLE     NOT NULL,
  `renewable_energy_pct`     DOUBLE     NOT NULL,
  `cost_usd`                 DOUBLE     NOT NULL,
  `delivery_time_days`       INT        NOT NULL,
  `sustainability_score`     DOUBLE     NOT NULL,
  `recyclability_score`      TINYINT(1) NOT NULL DEFAULT 0,
  `carbon_footprint_kg_co2`  INT        NOT NULL,
  `energy_efficiency_ratio`  DOUBLE     NOT NULL,
  `total_carbon_index`       DOUBLE     NOT NULL,
  `cost_per_km`              DOUBLE     NOT NULL,
  PRIMARY KEY (`fact_id`),
  KEY `idx_product_key`  (`product_key`),
  KEY `idx_delivery_key` (`delivery_key`),
  CONSTRAINT `fk_product`  FOREIGN KEY (`product_key`)  REFERENCES `dim_product`(`product_key`),
  CONSTRAINT `fk_delivery` FOREIGN KEY (`delivery_key`) REFERENCES `dim_delivery`(`delivery_key`)
) ENGINE=InnoDB DEFAULT CHARSET=utf8mb4;
""")
sql.append(df_to_inserts(fact_table, 'fact_supply_chain'))
sql.append('')

sql.append("""CREATE VIEW `v_supply_chain_full` AS
SELECT
  f.fact_id, p.product_type, p.sustainability_category,
  f.raw_material_usage_kg, f.energy_consumption_kwh,
  f.waste_generated_kg,    f.transport_distance_km,
  f.co2_emissions_kg,      f.manufacturing_energy_kwh,
  f.renewable_energy_pct,  f.cost_usd,
  f.delivery_time_days,    f.sustainability_score,
  f.recyclability_score,   f.carbon_footprint_kg_co2,
  f.energy_efficiency_ratio, f.total_carbon_index,
  f.cost_per_km
FROM fact_supply_chain f
JOIN dim_product  p ON f.product_key  = p.product_key
JOIN dim_delivery d ON f.delivery_key = d.delivery_key;
""")

with open('greenchain_db.sql', 'w', encoding='utf-8') as f:
    f.write('\n'.join(sql))

print('File greenchain_db.sql berhasil dibuat!')
print(f'   dim_product       : {len(dim_product)} baris')
print(f'   dim_delivery      : {len(dim_delivery)} baris')
print(f'   fact_supply_chain : {len(fact_table)} baris')
```

### Verifikasi Data Warehouse

```python
#VERIFIKASI DATA WAREHOUSE
print("=" * 55)
print("   VERIFIKASI DATA WAREHOUSE")
print("=" * 55)

verifikasi = []

# 1. Cek jumlah baris tiap tabel
verifikasi.append({
    'Tabel'       : 'dim_product',
    'Jumlah Baris': len(dim_product),
    'Status'      : 'OK' if len(dim_product) > 0 else '! KOSONG'
})
verifikasi.append({
    'Tabel'       : 'dim_delivery',
    'Jumlah Baris': len(dim_delivery),
    'Status'      : 'OK' if len(dim_delivery) > 0 else '! KOSONG'
})
verifikasi.append({
    'Tabel'       : 'fact_supply_chain',
    'Jumlah Baris': len(fact_table),
    'Status'      : 'OK' if len(fact_table) > 0 else '! KOSONG'
})

# 2. Cek tidak ada NULL di kolom kunci
null_product  = fact_table['product_key'].isnull().sum()
null_delivery = fact_table['delivery_key'].isnull().sum()
verifikasi.append({
    'Tabel'       : 'fact_supply_chain (product_key)',
    'Jumlah Baris': null_product,
    'Status'      : 'OK' if null_product == 0 else f'! {null_product} NULL'
})
verifikasi.append({
    'Tabel'       : 'fact_supply_chain (delivery_key)',
    'Jumlah Baris': null_delivery,
    'Status'      : 'OK' if null_delivery == 0 else f'! {null_delivery} NULL'
})

# 3. Cek foreign key — semua product_key di fact ada di dim_product
fk_product_invalid = ~fact_table['product_key'].isin(dim_product['product_key'])
verifikasi.append({
    'Tabel'       : 'FK: fact → dim_product',
    'Jumlah Baris': fk_product_invalid.sum(),
    'Status'      : 'OK' if fk_product_invalid.sum() == 0 else f'! {fk_product_invalid.sum()} tidak cocok'
})

# 4. Cek foreign key — semua delivery_key di fact ada di dim_delivery
fk_delivery_invalid = ~fact_table['delivery_key'].isin(dim_delivery['delivery_key'])
verifikasi.append({
    'Tabel'       : 'FK: fact → dim_delivery',
    'Jumlah Baris': fk_delivery_invalid.sum(),
    'Status'      : 'OK' if fk_delivery_invalid.sum() == 0 else f'! {fk_delivery_invalid.sum()} tidak cocok'
})

# 5. Cek duplikat di fact_id
dup_fact = fact_table['fact_id'].duplicated().sum()
verifikasi.append({
    'Tabel'       : 'fact_supply_chain (fact_id duplikat)',
    'Jumlah Baris': dup_fact,
    'Status'      : 'OK' if dup_fact == 0 else f'! {dup_fact} duplikat'
})

# 6. Cek konsistensi jumlah
selisih = len(fact_table) - len(df_clean)
verifikasi.append({
    'Tabel'       : 'Konsistensi fact vs df_clean',
    'Jumlah Baris': abs(selisih),
    'Status'      : 'OK' if selisih == 0 else f'! selisih {selisih} baris'
})

# Tampilkan hasil
df_verif = pd.DataFrame(verifikasi)
print(df_verif.to_string(index=False))

# Ringkasan akhir
ada_masalah = df_verif[~df_verif['Status'].str.startswith('OK')]
print(f"\n{'='*55}")
if ada_masalah.empty:
    print("Verifikasi Berhasil, Data Warehouse siap digunakan!")
else:
    print(f"! Ditemukan {len(ada_masalah)} item bermasalah — cek tabel di atas.")
print("=" * 55)
```
